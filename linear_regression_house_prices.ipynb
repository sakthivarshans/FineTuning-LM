## Cell 1: Import libraries and load data

```
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df = pd.read_csv("house_prices_raw.csv")
df.head()
```

## Cell 2: Check the data

```
print(df.shape)
print(df.isna().sum())
print("Duplicates:", df.duplicated().sum())
```

## Cell 3: Clean the data

```
df = df.drop_duplicates()                              # remove duplicates
df["City"] = df["City"].str.strip().str.title()        # fix city names
df = df.dropna()                                       # remove missing values
df = df[df["Age_years"] >= 0]                          # remove negative age
df = df[df["Price"] < df["Price"].quantile(0.98)]      # remove very high prices
df = df[df["Area_sqft"] < df["Area_sqft"].quantile(0.98)]  # remove very large areas

print(df.shape)
print(df.isna().sum().sum(), "missing values left")
```

## Cell 4: Chart 1 - Histogram

```
plt.hist(df["Price"], bins=20)
plt.title("Price Histogram")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

## Cell 5: Chart 2 - Bar chart

```
avg = df.groupby("City")["Price"].mean()

plt.bar(avg.index, avg.values)
plt.title("Average Price by City")
plt.xlabel("City")
plt.ylabel("Average Price")
plt.show()
```

## Cell 6: Linear Regression

```
X = df[["Area_sqft"]]
y = df["Price"]

model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
print("R2 score:", r2_score(y, model.predict(X)))
```

## Cell 7: Plot the regression line

```
plt.scatter(X, y, label="Data")
plt.plot(X, model.predict(X), color="red", label="Regression line")
plt.title("Area vs Price")
plt.xlabel("Area (sqft)")
plt.ylabel("Price")
plt.legend()
plt.show()
```
